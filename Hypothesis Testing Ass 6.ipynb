```python
import scipy.stats as stats
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
## from the given data we can find the theorotical mean of the operating model
x=600
w=1000+(5*x)
Theo_mean=w
Theo_mean
```

```python
n=25
pop_std_dev=25
population_std_dev=abs(5)*pop_std_dev
population_std_dev
```

```python
### next finding the error
error=population_std_dev/np.sqrt(n)
error
```

```python
### finding the z-score of the model using the values calculated above
sample_mean=3050
Theo_mean=4000
z_score=(sample_mean-Theo_mean)/error
z_score
```

#### Hence, we can see that there is diffrence between the sample mean which is 3050 and the theorotical mean calculated as 4000 which results in 
z_score of -38.0
As the sample mean is lower than the theorotical mean 
Null Hypothesisi(H0) is proven to be true and selected
and Alternate Hypothesis(Ha) is rejected
