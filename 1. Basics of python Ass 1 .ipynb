```python

num=int(input('Enter the number:'))
if num>1:
    ### checking for factors
    for i in range(2,num):
       if(num % i)==0:
         print(num,"is not prime number")
         break
    else:
        print(num,"is prime number")
else:
        print(num,"is not a prime number")
    
        
```

```python
### given
room_charges_per_day=2000
consultation_fees=1500
cost_per_lab_test=300
days_stayed=4
test_done=3
medicine_charges=2400
### 10% discount incase bill exceed 10000
```

```python
### bill before applying discount
total_room_charges=(room_charges_per_day)*(days_stayed)
total_lab_charges = (cost_per_lab_test) * (test_done)
bill_before_applying_discount = total_room_charges + consultation_fees + total_lab_charges + medicine_charges
a=bill_before_applying_discount
print("The bill before applying discount is:",a)

# Apply discount logic
if a > 10000:
    dis_rate = 0.10
    discount_amount = a * dis_rate
    final_bill = a - discount_amount
else:
    final_bill = a

print(f"The patient's final bill is ₹{final_bill}")
```
